```python
from random import randint
computador = randint(1, 10)
print('sou o computador e vou pensar em um numero de 1 a 10, você terá que advinhar qual foi!')
acertou = False
palpite = 0
while not acertou:    #enquanto NÃO acertou, ou seja, nesse caso, enquanto não for (NÃO FALSE) haverá repetição
  jogador = int(input('qual o seu palpite? '))
  palpite += 1
  if jogador == computador:
    acertou = True
  elif jogador > computador:
    print('um pouco menor, tente novamente')
  elif jogador < computador:
    print('um pouco maior, tente novamente')
print(f'após {palpite} tentativas, você acertou!')
```

```python
from time import sleep
def contador(i, f , p):
  print(f'contagem de {i} ate {f} de {p} em {p} ')

  if i < f:
    cont = i
    while cont <= f:
      print(f'{cont} ', end='', flush=True)
      sleep(0.5)
      cont += p
    print('Fim!')
  else:
    cont = i
    while cont >= f:
      print(f'{cont} ', end='', flush=True)
      sleep(0.5)
      cont -= p
    print('fim!')

C = int(input('começo da contagem: '))
D = int(input('fim da contagem: '))
P = int(input('passo: '))
contador(C, D, P)
```
